# Kaggle: O que faz um vídeo viralizar? (12.000 vídeos)

```python
import pandas as pd

# 1. Carregar os dados
df = pd.read_csv('viral_videos.csv')
df.head(3)
```

```python
# 2. Limpar e Renomear
df.rename(columns={'category': 'categoria', 'views': 'visualizacoes', 'likes': 'curtidas'}, inplace=True)
```

```python
# 3. Filtrar vídeos de Sucesso (> 500k views) em Tech
df_sucesso = df[(df['visualizacoes'] > 500000) & (df['categoria'] == 'Tech')]
print('Quantidade de vídeos de Tech que viralizaram:', len(df_sucesso))
```

```python
# 4. O que faz viralizar? Rosto na Thumbnail
print('Média de visualizações (0 = Sem Rosto, 1 = Com Rosto):')
df.groupby('thumbnail_face')['visualizacoes'].mean()
```

```python
# 5. Melhor dia para postar (soma de curtidas)

# Dicionário de tradução
dias_br = {
    'Mon': 'Segunda', 'Tue': 'Terça', 'Wed': 'Quarta', 
    'Thu': 'Quinta', 'Fri': 'Sexta', 'Sat': 'Sábado', 'Sun': 'Domingo'
}

# Aplicando a tradução na coluna
df['upload_day'] = df['upload_day'].replace(dias_br)

# Fazendo o agrupamento com os nomes em português
print('Total de curtidas por dia da semana:')
df.groupby('upload_day')['curtidas'].sum().sort_values(ascending=False)
```
